```python
#assignment operator
#=, +=, -=, *=, /=, //=, %=, **=


a = 25
b = 7
c = 2
d = b
print("identifier of \"a\"->",id(a))
print("identifier of \"b\"->",id(b))
print("identifier of \"c\"->",id(c))
print("identifier of \"d\"->",id(d))
a += 5      #a=30
print("value of a -",a)
b -= 7      #b=0
print("value of b -",b,id(b))
c *= 5      #c=10
print("value of c -",c)
d /=3       #d=2.3              #HERE THE VALUE OF D IS 7 NOT ZERO BECAUSE THE D=B=7 THE D IS POINTING THE VALUE 7 IN THE MEMORY, NOT B 
print("value of d -",d,id(d))
a //=10     #a=3
print("value of a -",a)
c %=3       #c=1
print("value of c -",c)
a **=3      #a=27
print("value of a -",a)
```

```python
# COMPRISON OPERATOR
# ==, !=, >, <, >=, <=

print(24 == 24) #T
print(24 != 24) #F
print(24 > 24)  #F
print(24 >= 24) #T
print(24 < 24)  #F
print(24 <= 24) #T
```
